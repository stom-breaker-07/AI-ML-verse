```python
def Even_Odd(x):
    return True if x%2==0 else False

print(Even_Odd(7))
```

```python
# Finding the max in a list
numbers = [4, 17, 92, 5, 23]
largest = max(numbers)
print(largest)  # Outputs: 92
```

```python
# Finding the max between independent variables
a = 15
b = 42
c = 7
largest_var = max(a, b, c)
print(largest_var)  # Outputs: 42
```

```python
# Gusse Game 
import random as r

print("Welcome to guessing game ")
ran=r.randint(1,10)
choose=0
chance=0

while choose!=ran and chance < 5:
    choose=int(input("Gusse the number btwn 1-10 : "))
    chance+=1;

if chance > 5:
    print("You Lost the Chance")
else :
    print("You Gusse correct number ",ran ,"and in " ,chance, "number of chance")
```
