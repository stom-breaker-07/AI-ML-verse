```python
exit=True

print ("Welcome to TODO List")
List=[]

def add(str) :
    return List.append(str)

def remove(idx) :
    return List.pop(idx)

def viewAll() :
    print("Printing all Tasks")
    for task in List :
        print(task)


while True :
    choice=int(input("1.Add Task \n 2.Remove Task \n 3.View All List \n 4.Exit "))
    match choice :
        case 1:
            task=input("Enter the Task : " )
            add(task) 
            print("Task added Successfully")
        case 2:
            index=int(input("Enter the index"))
            remove(index)
            print("Removed Successfully")
        case 3:
            viewAll()
        case 4:
            break
        case _:
            print("Incorrect Option")


            
```
