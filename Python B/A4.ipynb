```python
# dictonary 
student=[]
def Add(d):
    student.append(d)
    print("Added Succussfully")

def Show():
    print("Data Showing ...")
    for item in student:
        print("Name :" ,item["name"] ,"Age :" ,item["age"])

def Update(index):
    dic=student[index]
    col=input("Enter the Column name :")
    value=input("Enter the Value :")
    dic[col.lower()]=value
    print("Value Updated Successfully")
    
print("Welcome to Marks Analysis Sheet")

while True:
    print(''' 1.Add Studwnt Details \n 2.Show Student Details \n 3.Update Student Details \n 4.Exit''')
    choice=int(input("Enter the Option :"))
    
    match choice:
        case 1:
            n=input("Enter the name :")
            a=input("Enter the age :")
            Add({"name":n,"age":a})
        case 2:
            Show()
        case 3:
            idx=int(input("Enter the Row of List"))
            Update(idx)
        case 4:
            print("Exiting Successfully")
            break
        case _:
            print("Incorrect Option ")
```
