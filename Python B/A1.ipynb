```python
# Project: Calculator with Match-Case

def Sum(a, b):
    return a + b

def Diff(a, b):
    return a - b

def Prod(a, b):
    return a * b

def Div(a, b):
    if b == 0:
        return "Error: Cannot divide by zero!"
    return a / b

def FloorDiv(a, b):
    if b == 0:
        return "Error: Cannot divide by zero!"
    return a // b

def ToThePow(a, b):
    return a ** b

# Loop remains true until explicitly set to False or broken
loop = True

while loop:
    # Added spaces inside inputs for better readability in terminal
    a = int(input("First Number: "))
    b = int(input("Second Number: "))
    choice = input("Enter Choice (Add, Sub, Mul, Div, FDiv, Pow): ")

    match choice:
        case "Add":
            print("Sum is:", Sum(a, b))
        case "Sub":
            print("Sub is:", Diff(a, b))
        case "Mul":
            print("Prod is:", Prod(a, b))
        case "Div":
            print("Div is:", Div(a, b))
        case "FDiv":
            print("Floor Div is:", FloorDiv(a, b))
        case "Pow":
            print("To the Power is:", ToThePow(a, b))
        case _:
            print("Invalid Choice. Exiting...")
            loop = False  # Properly terminates the while loop
```
